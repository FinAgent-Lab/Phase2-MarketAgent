```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
```

```python
def openrouter_chat(prompt, model="openai/gpt-4o-mini"):
    """
    OpenRouter로 LLM 응답 받는 헬퍼 함수
    model 예시:
      - openai/gpt-4o-mini
      - anthropic/claude-3.5-sonnet
      - google/gemini-pro-1.5
    """
    url = "https://openrouter.ai/api/v1/chat/completions"

    headers = {
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "Content-Type": "application/json",
        "HTTP-Referer": "http://localhost",
        "X-Title": "Stock Agent",
    }

    payload = {
        "model": model,
        "messages": [
            {"role": "user", "content": prompt}
        ]
    }

    response = requests.post(url, headers=headers, json=payload)
    data = response.json()
    return data["choices"][0]["message"]["content"]
```

```python
openrouter_chat("안녕, 너는 지금 무슨 모델인지 짧게 설명해줄래?")
```

```python
import requests
import pandas as pd
import os

ALPHA_KEY = os.getenv("ALPHAVANTAGE_API_KEY")

def fetch_alpha_daily(symbol="AAPL"):
    url = (
        "https://www.alphavantage.co/query"
        f"?function=TIME_SERIES_DAILY&symbol={symbol}"
        f"&apikey={ALPHA_KEY}"
    )
    data = requests.get(url).json()
    return data

raw = fetch_alpha_daily("AAPL")
```

```python
raw
```

```python
daily = raw.get("Time Series (Daily)", {})
daily
```

```python
def alpha_to_df(data):
    daily = data.get("Time Series (Daily)", {})
    df = pd.DataFrame.from_dict(daily, orient="index")
    df.columns = ["open","high","low","close","volume"]
    df.index = pd.to_datetime(df.index)
    df = df.sort_index()
    return df.astype(float)

df = alpha_to_df(raw)
df.tail()
```

```python
recent_df = df.tail(20)
```

```python
closes = "\n".join([
    f"{i.strftime('%Y-%m-%d')}: {row['close']}"
    for i, row in recent_df.iterrows()
])
closes
```

```python
prompt = f"""
너는 미국 주식 전문 애널리스트이다.
티커: AAPL
최근 20일 종가 데이터:
{closes}

다음 내용을 분석해줘:
- 추세 분석
- 변동성 평가
- 주요 리스크
- 단기 방향성 전망
"""
```
