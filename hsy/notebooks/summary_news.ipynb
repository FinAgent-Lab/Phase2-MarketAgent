# 데일리 뉴스 리포트 요약기

이 노트북은 `daily_reports` 폴더에 저장된 마크다운 리포트 파일을 읽어 LLM을 통해 요약 정보를 생성합니다.

```python
# 1. 환경 설정 및 라이브러리 임포트
import os
import sys
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

# 현재 노트북의 상위 경로를 sys.path에 추가 (모듈 import 등을 위해)
sys.path.append("..")

# .env 파일 로드 (상위 디렉토리에 위치)
load_dotenv("../.env")

# LLM 초기화
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

llm = ChatOpenAI(
    model="tngtech/deepseek-r1t2-chimera:free",
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1",
)
```

```python
# 2. 리포트 읽기 및 요약 함수 정의
def summarize_daily_report(target_date: str):
    """
    특정 날짜의 daily_report 파일을 읽어서 LLM을 이용해 요약합니다.
    
    Args:
        target_date (str): 'YYYY-MM-DD' 형식의 날짜 문자열
    """
    # 파일 경로 구성
    # notebooks 폴더 상위의 daily_reports 폴더를 참조
    file_path = f"../daily_reports/daily_report_{target_date}.md"
    
    # 파일 존재 여부 확인
    if not os.path.exists(file_path):
        print(f"Error: 해당 날짜의 리포트 파일이 존재하지 않습니다. ({file_path})")
        return

    # 파일 읽기
    try:
        with open(file_path, "r", encoding="utf-8") as f:
            content = f.read()
            
        print(f"파일을 성공적으로 읽었습니다: {file_path}")
        print(f"문서 길이: {len(content)} 자")
        
    except Exception as e:
        print(f"파일 읽기 중 오류 발생: {e}")
        return

    # LLM 요약 요청
    print("\n--- 요약 생성 중 ---\n")
    
    prompt = (
        f"당신은 숙련된 투자 분석가입니다. 아래는 {target_date}의 주식 및 경제 뉴스 리포트입니다.\n"
        "이 리포트의 내용을 바탕으로 다음 사항들을 포함하여 핵심을 요약해주세요:\n\n"
        "1. **오늘의 핵심 이슈**: 가장 중요한 시장 이벤트나 뉴스 3가지\n"
        "2. **시장 동향 요약**: 주요 섹터별 흐름 및 특이사항\n"
        "3. **투자 시사점**: 투자자가 주목해야 할 포인트\n\n"
        "리포트 내용:\n"
        f"{content}\n\n"
        "응답은 보기 좋은 마크다운 형식으로 작성해주세요."
    )
    
    try:
        response = llm.invoke([{"role": "user", "content": prompt}])
        from IPython.display import display, Markdown
        display(Markdown(response.content))
    except Exception as e:
        print(f"LLM 요청 중 오류 발생: {e}")
```

```python
# 3. 실행
# 요약할 날짜를 입력하세요 (예: '2025-12-11')
target_date = input("날짜 입력 (YYYY-MM-DD): ")
if not target_date:
    target_date = "2025-12-11"

summarize_daily_report(target_date)
```
